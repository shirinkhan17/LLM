# What Is Context Window?

Context window means:

```text
how many previous tokens the model can see at once
```

during attention.

---

# Example

Sentence:

```text
The patient received medicine at the hospital.
```

Each word/subword becomes tokens.

Transformer attends over:

```text
previous context tokens
```

---

# GPT-2

GPT-2 can only look at:

```text
last 1024 tokens
```

Maximum.

Anything older:
- disappears from attention
- cannot be directly used.

---

# GPT-3

GPT-3 can look at:

```text
last 2048 tokens
```

So:
- longer documents
- bigger prompts
- more examples

fit inside model memory.

---

````python
# pip install transformers torch

import torch
from transformers import AutoTokenizer, AutoModelForCausalLM

# Use GPT-2 as a small public GPT-style model
MODEL_NAME = "gpt2"

tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)
tokenizer.pad_token = tokenizer.eos_token

model = AutoModelForCausalLM.from_pretrained(MODEL_NAME)
model.config.pad_token_id = tokenizer.eos_token_id

# Few-shot prompt like GPT-3 paper
prompt = """
English: cat
French: chat

English: dog
French: chien

English: house
French:
"""

inputs = tokenizer(prompt, return_tensors="pt")

with torch.no_grad():
    output_ids = model.generate(
        input_ids=inputs["input_ids"],
        attention_mask=inputs["attention_mask"],
        max_new_tokens=10,
        do_sample=False,
        pad_token_id=tokenizer.eos_token_id
    )

output = tokenizer.decode(output_ids[0], skip_special_tokens=True)

print(output)
````
