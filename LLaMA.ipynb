````python
import torch
import torch.nn as nn
import torch.nn.functional as F
import math

torch.manual_seed(42)

# =====================================================
# 1. Tiny dataset
# =====================================================

texts = [
    "the cat sat on mat",
    "the dog ran in park",
    "patient received medicine",
    "bank approved loan"
]

special_tokens = ["<pad>", "<bos>", "<eos>"]

all_words = set(special_tokens)

for text in texts:
    all_words.update(text.split())

vocab = {w: i for i, w in enumerate(sorted(all_words))}
id_to_word = {i: w for w, i in vocab.items()}

PAD = vocab["<pad>"]
BOS = vocab["<bos>"]
EOS = vocab["<eos>"]

def encode(text, max_len=8):
    ids = [BOS] + [vocab[w] for w in text.split()] + [EOS]
    ids = ids[:max_len]
    ids += [PAD] * (max_len - len(ids))
    return ids

max_len = 8

data = torch.tensor([
    encode(text, max_len)
    for text in texts
])

input_ids = data[:, :-1]
target_ids = data[:, 1:]

# =====================================================
# 2. RMSNorm
# =====================================================

class RMSNorm(nn.Module):
    def __init__(self, d_model, eps=1e-6):
        super().__init__()
        self.weight = nn.Parameter(torch.ones(d_model))
        self.eps = eps

    def forward(self, x):
        rms = torch.sqrt(torch.mean(x * x, dim=-1, keepdim=True) + self.eps)
        return self.weight * (x / rms)

# =====================================================
# 3. RoPE
# =====================================================

def apply_rope(x):
    batch, heads, seq_len, head_dim = x.shape

    half_dim = head_dim // 2

    x1 = x[..., :half_dim]
    x2 = x[..., half_dim:]

    positions = torch.arange(seq_len, device=x.device).float()

    freqs = torch.arange(half_dim, device=x.device).float()
    freqs = 1.0 / (10000 ** (freqs / half_dim))

    angles = positions[:, None] * freqs[None, :]

    sin = torch.sin(angles)[None, None, :, :]
    cos = torch.cos(angles)[None, None, :, :]

    rotated = torch.cat(
        [
            x1 * cos - x2 * sin,
            x1 * sin + x2 * cos
        ],
        dim=-1
    )

    return rotated

# =====================================================
# 4. Causal Self-Attention with RoPE
# =====================================================

class CausalSelfAttention(nn.Module):
    def __init__(self, d_model=64, num_heads=4):
        super().__init__()

        self.num_heads = num_heads
        self.head_dim = d_model // num_heads

        self.q = nn.Linear(d_model, d_model, bias=False)
        self.k = nn.Linear(d_model, d_model, bias=False)
        self.v = nn.Linear(d_model, d_model, bias=False)
        self.o = nn.Linear(d_model, d_model, bias=False)

    def forward(self, x):
        B, T, C = x.shape

        q = self.q(x)
        k = self.k(x)
        v = self.v(x)

        q = q.view(B, T, self.num_heads, self.head_dim).transpose(1, 2)
        k = k.view(B, T, self.num_heads, self.head_dim).transpose(1, 2)
        v = v.view(B, T, self.num_heads, self.head_dim).transpose(1, 2)

        # LLaMA-style RoPE applied to Q and K
        q = apply_rope(q)
        k = apply_rope(k)

        scores = torch.matmul(q, k.transpose(-2, -1))
        scores = scores / math.sqrt(self.head_dim)

        causal_mask = torch.triu(
            torch.ones(T, T, device=x.device),
            diagonal=1
        ).bool()

        scores = scores.masked_fill(causal_mask, float("-inf"))

        attn = F.softmax(scores, dim=-1)

        out = torch.matmul(attn, v)

        out = out.transpose(1, 2).contiguous().view(B, T, C)

        return self.o(out)

# =====================================================
# 5. SwiGLU Feedforward
# =====================================================

class SwiGLU(nn.Module):
    def __init__(self, d_model=64, hidden_dim=128):
        super().__init__()

        self.w1 = nn.Linear(d_model, hidden_dim, bias=False)
        self.w2 = nn.Linear(d_model, hidden_dim, bias=False)
        self.w3 = nn.Linear(hidden_dim, d_model, bias=False)

    def forward(self, x):
        return self.w3(F.silu(self.w1(x)) * self.w2(x))

# =====================================================
# 6. Tiny LLaMA Block
# =====================================================

class TinyLLaMABlock(nn.Module):
    def __init__(self, d_model=64, num_heads=4):
        super().__init__()

        self.norm1 = RMSNorm(d_model)
        self.attn = CausalSelfAttention(d_model, num_heads)

        self.norm2 = RMSNorm(d_model)
        self.ffn = SwiGLU(d_model, hidden_dim=128)

    def forward(self, x):
        x = x + self.attn(self.norm1(x))
        x = x + self.ffn(self.norm2(x))
        return x

# =====================================================
# 7. Tiny LLaMA Model
# =====================================================

class TinyLLaMA(nn.Module):
    def __init__(self, vocab_size, d_model=64, num_heads=4, num_layers=2):
        super().__init__()

        self.token_embedding = nn.Embedding(vocab_size, d_model)

        self.layers = nn.ModuleList([
            TinyLLaMABlock(d_model, num_heads)
            for _ in range(num_layers)
        ])

        self.norm = RMSNorm(d_model)
        self.lm_head = nn.Linear(d_model, vocab_size, bias=False)

    def forward(self, input_ids):
        x = self.token_embedding(input_ids)

        for layer in self.layers:
            x = layer(x)

        x = self.norm(x)

        logits = self.lm_head(x)

        return logits

# =====================================================
# 8. Train
# =====================================================

model = TinyLLaMA(vocab_size=len(vocab))

optimizer = torch.optim.AdamW(model.parameters(), lr=0.01)
loss_fn = nn.CrossEntropyLoss(ignore_index=PAD)

for epoch in range(300):
    logits = model(input_ids)

    loss = loss_fn(
        logits.reshape(-1, len(vocab)),
        target_ids.reshape(-1)
    )

    optimizer.zero_grad()
    loss.backward()
    optimizer.step()

    if (epoch + 1) % 50 == 0:
        print(f"Epoch {epoch+1}, Loss={loss.item():.4f}")

# =====================================================
# 9. Generate
# =====================================================

def generate(prompt, max_new_tokens=5):
    model.eval()

    ids = [BOS] + [vocab.get(w, PAD) for w in prompt.split()]

    for _ in range(max_new_tokens):
        x = torch.tensor([ids])

        with torch.no_grad():
            logits = model(x)

        next_id = logits[:, -1, :].argmax(dim=-1).item()

        if next_id == EOS:
            break

        ids.append(next_id)

    words = [
        id_to_word[i]
        for i in ids
        if i not in [BOS, EOS, PAD]
    ]

    return " ".join(words)

print("\nGenerated:")
print(generate("the cat"))
````

````python
# =====================================================
# 9. Rule-Based Inference / Generate
# =====================================================

def inference_with_rule(prompt, allowed_words, max_new_tokens=5):
    model.eval()

    ids = [BOS] + [vocab.get(w, PAD) for w in prompt.split()]

    allowed_ids = [
        vocab[w]
        for w in allowed_words
        if w in vocab
    ]

    print("\nInitial Prompt:", prompt)
    print("Allowed next words:", allowed_words)

    for step in range(max_new_tokens):
        x = torch.tensor([ids])

        with torch.no_grad():
            logits = model(x)

        next_logits = logits[:, -1, :]

        # rule: block all words except allowed words
        mask = torch.full_like(next_logits, float("-inf"))
        mask[:, allowed_ids] = next_logits[:, allowed_ids]

        next_id = mask.argmax(dim=-1).item()

        if next_id == EOS:
            break

        ids.append(next_id)

        print(f"Step {step+1} predicted:", id_to_word[next_id])

    words = [
        id_to_word[i]
        for i in ids
        if i not in [BOS, EOS, PAD]
    ]

    return " ".join(words)
````

````python
from transformers import AutoTokenizer, AutoModelForCausalLM
import torch

MODEL_NAME = "TinyLlama/TinyLlama-1.1B-Chat-v1.0"

tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)

model = AutoModelForCausalLM.from_pretrained(
    MODEL_NAME,
    torch_dtype=torch.float16,
    device_map="auto"
)

# -------------------------------------------------
# Few-shot prompt
# -------------------------------------------------

prompt = """Complete the pattern.

Input: 2
Output: 4

Input: 3
Output: 6

Input: 5
Output: 10

Input: 7
Output:"""


inputs = tokenizer(prompt, return_tensors="pt").to(model.device)

with torch.no_grad():
    output_ids = model.generate(
        **inputs,
        max_new_tokens=3,
        do_sample=False,
        pad_token_id=tokenizer.eos_token_id,
        eos_token_id=tokenizer.eos_token_id,
    )

new_tokens = output_ids[0][inputs["input_ids"].shape[-1]:]
print(tokenizer.decode(new_tokens, skip_special_tokens=True).strip())
````
