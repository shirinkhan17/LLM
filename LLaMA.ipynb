| Feature | GPT (2018) | GPT-2 (2019) | GPT-3 (2020) | InstructGPT (2022) | LLaMA (2023) |
|---|---|---|---|---|---|
| Paper | *Improving Language Understanding by Generative Pre-Training* | *Language Models are Unsupervised Multitask Learners* | *Language Models are Few-Shot Learners* | *Training Language Models to Follow Instructions with Human Feedback* | *Open and Efficient Foundation Language Models* |
| Organization | OpenAI | OpenAI | OpenAI | OpenAI | Meta AI |
| Architecture | decoder-only Transformer | larger decoder-only Transformer | massive decoder-only Transformer | GPT-3 + RLHF | optimized decoder-only Transformer |
| Training Objective | next-token prediction | next-token prediction | next-token prediction | instruction tuning + RLHF | next-token prediction |
| Attention | causal self-attention | causal self-attention | causal self-attention | causal self-attention | causal self-attention |
| Direction | left-to-right | left-to-right | left-to-right | left-to-right | left-to-right |
| Main Goal | generative pretraining | large-scale generation | few-shot learning | instruction following | efficient open LLM |
| Parameters | 117M | up to 1.5B | up to 175B | GPT-3 based | 7B–65B |
| Context Window | 512 | 1024 | 2048 | GPT-3 based | 2048+ |
| Training Data | BooksCorpus | WebText | massive internet corpus | GPT-3 + human feedback | 1T+ tokens |
| Fine-Tuning Style | supervised downstream fine-tuning | mostly zero-shot | in-context learning | supervised + PPO RLHF | instruction tuning capable |
| Zero-Shot Ability | weak | moderate | strong | very strong | strong |
| Few-Shot Ability | limited | improved | powerful | powerful + aligned | powerful |
| Positional Embedding | learned absolute | learned absolute | learned absolute | learned absolute | RoPE |
| Normalization | Pre-Norm + LayerNorm | Pre-Norm + LayerNorm | Pre-Norm + LayerNorm | Pre-Norm + LayerNorm | Pre-Norm + RMSNorm |
| Feedforward Activation | GELU | GELU | GELU | GELU | SwiGLU |
| Key Innovation | pretrained LM | scaling generation | emergent few-shot learning | RLHF alignment | efficient scaling |
| Training Paradigm | pretrain → fine-tune | large-scale LM | prompt-based learning | human-aligned learning | efficient open-source LLM |
| Emergent Abilities | limited | better generation | reasoning + in-context learning | safer/helpful responses | smaller but powerful models |
| RLHF | no | no | no | yes | optional in derived models |
| Open Source | partially | limited | closed | closed | research-open |
| Impact | introduced GPT family | showed scaling power | launched LLM era | foundation of ChatGPT | accelerated open LLM ecosystem |

# Evolution Summary

```text
GPT
→ generative pretraining

GPT-2
→ scaling improves generation

GPT-3
→ scaling creates emergent few-shot learning

InstructGPT
→ RLHF aligns models with human instructions

LLaMA
→ efficient modern open foundation models
```
Both are:

```text
decoder-only Transformers
```

trained with:

```text
next-token prediction
```

BUT LLaMA introduced several newer architectural improvements and training strategies.

---

# Core Difference

| GPT-3 | LLaMA |
|---|---|
| larger scaling focus | efficiency + scaling |
| LayerNorm | RMSNorm |
| GELU FFN | SwiGLU FFN |
| absolute positional embeddings | RoPE |
| older Transformer design | optimized modern design |
| 175B parameters | smaller but competitive |

---

# 1. Positional Embeddings

## GPT-3

Uses:

```text
learned absolute positional embeddings
```

---

# LLaMA

Uses:

```text
RoPE (Rotary Positional Embeddings)
```

Benefits:
- better long-context handling
- relative position awareness
- extrapolation to longer contexts.

---

# 2. Normalization

## GPT-3

Uses:

```text
LayerNorm
```

---

# LLaMA

Uses:

```text
RMSNorm
```

Benefits:
- simpler
- faster
- more stable.

---

# 3. Feedforward Activation

## GPT-3

Uses:

```text
GELU
```

---

# LLaMA

Uses:

```text
SwiGLU
```

Benefits:
- stronger learning capacity
- better gradient flow.

---

# 4. Efficiency

LLaMA showed:

```text
smaller models can compete with larger ones
```

Example:

```text
LLaMA-13B ≈ GPT-3 175B
```

on several benchmarks.

---

# 5. Training Data Strategy

## GPT-3

Focused mainly on:
- huge parameter scaling.

---

# LLaMA

Focused on:
- more training tokens
- better scaling laws
- compute efficiency.

Inspired by:
- Chinchilla scaling.

---

# 6. Open vs Closed

## GPT-3

Mostly closed-source.

---

# LLaMA

Released for research/open ecosystem,
leading to:
- Alpaca
- Vicuna

- many open LLMs.
