````python
# pip install transformers torch

import torch
from transformers import AutoTokenizer, AutoModelForCausalLM

# -------------------------------------------------
# 1. Load pretrained GPT-2
# -------------------------------------------------

MODEL_NAME = "gpt2"

tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)

model = AutoModelForCausalLM.from_pretrained(MODEL_NAME)

tokenizer.pad_token = tokenizer.eos_token
model.config.pad_token_id = tokenizer.eos_token_id

# -------------------------------------------------
# 2. Zero-shot prompting (NO fine-tuning)
# -------------------------------------------------

prompt = """
Article:
The patient was diagnosed with diabetes and received medicine at the hospital.

Summary:
"""

inputs = tokenizer(
    prompt,
    return_tensors="pt"
)

# -------------------------------------------------
# 3. Generate directly from pretrained GPT-2
# -------------------------------------------------

with torch.no_grad():

    output_ids = model.generate(
        input_ids=inputs["input_ids"],
        attention_mask=inputs["attention_mask"],
        max_new_tokens=20,
        do_sample=True,
        temperature=0.7,
        top_k=50,
        top_p=0.95,
        no_repeat_ngram_size=2,
        repetition_penalty=1.5,
        pad_token_id=tokenizer.eos_token_id
    )

output = tokenizer.decode(
    output_ids[0],
    skip_special_tokens=True
)

print(output)
````

GPT2 alone is not good for summrization, it requires fine tuning, but GPT2 can be used without fine tuning, only using promt.
