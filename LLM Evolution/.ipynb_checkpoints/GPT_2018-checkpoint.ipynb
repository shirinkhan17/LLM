```python
import torch
import torch.nn as nn
import torch.nn.functional as F
import math

torch.manual_seed(42)

# -----------------------------
# 1. Tiny pretraining corpus
# -----------------------------
corpus = [
    "i love machine learning",
    "i love deep learning",
    "machine learning is useful",
    "deep learning is powerful",
    "i like natural language processing",
    "language models predict next word",
]

# -----------------------------
# 2. Build vocabulary
# -----------------------------
words = sorted(set(" ".join(corpus).split()))
vocab = {"<PAD>": 0, "<UNK>": 1}
for w in words:
    vocab[w] = len(vocab)

id_to_word = {i: w for w, i in vocab.items()}
vocab_size = len(vocab)

def encode(text):
    return [vocab.get(w, vocab["<UNK>"]) for w in text.split()]

# -----------------------------
# 3. Create pretraining data
# -----------------------------
seq_len = 4

training_sequences = []

for sentence in corpus:
    ids = encode(sentence)

    for i in range(1, len(ids)):
        context = ids[max(0, i-seq_len):i]
        target = ids[i]

        context = [0] * (seq_len - len(context)) + context
        training_sequences.append((context, target))

X = torch.tensor([x for x, y in training_sequences])
y = torch.tensor([y for x, y in training_sequences])

print("Pretraining input shape:", X.shape)
print("Pretraining target shape:", y.shape)
```

```python
class TinyGPT(nn.Module):
    def __init__(self, vocab_size, d_model=32, num_heads=4, num_layers=2, max_len=4):
        super().__init__()

        self.token_embedding = nn.Embedding(vocab_size, d_model)
        self.position_embedding = nn.Embedding(max_len, d_model)

        encoder_layer = nn.TransformerEncoderLayer(
            d_model=d_model,
            nhead=num_heads,
            dim_feedforward=64,
            batch_first=True
        )

        self.transformer = nn.TransformerEncoder(
            encoder_layer,
            num_layers=num_layers
        )

        self.lm_head = nn.Linear(d_model, vocab_size)
        self.max_len = max_len

    def causal_mask(self, size):
        mask = torch.triu(torch.ones(size, size), diagonal=1).bool()
        return mask

    def forward(self, input_ids):
        batch_size, seq_len = input_ids.shape

        positions = torch.arange(seq_len).unsqueeze(0).expand(batch_size, seq_len)

        x = self.token_embedding(input_ids) + self.position_embedding(positions)

        mask = self.causal_mask(seq_len)

        x = self.transformer(x, mask=mask)

        final_hidden = x[:, -1, :]

        logits = self.lm_head(final_hidden)

        return logits
```

```python
model = TinyGPT(vocab_size=vocab_size)

optimizer = torch.optim.Adam(model.parameters(), lr=0.01)
loss_fn = nn.CrossEntropyLoss()

for epoch in range(300):
    logits = model(X)
    loss = loss_fn(logits, y)

    optimizer.zero_grad()
    loss.backward()
    optimizer.step()

    if (epoch + 1) % 50 == 0:
        pred = logits.argmax(dim=1)
        acc = (pred == y).float().mean().item()
        print(f"Pretrain Epoch {epoch+1}, Loss: {loss.item():.4f}, Acc: {acc:.2f}")
```

```python
def predict_next(text):
    ids = encode(text)
    ids = ids[-seq_len:]
    ids = [0] * (seq_len - len(ids)) + ids

    x = torch.tensor([ids])

    with torch.no_grad():
        logits = model(x)

    pred_id = logits.argmax(dim=1).item()
    return id_to_word[pred_id]

print(predict_next("i love machine"))
print(predict_next("deep learning is"))
```

```python
fine_tune_data = [
    ("i love machine learning", 1),
    ("deep learning is powerful", 1),
    ("language models are useful", 1),
    ("machine learning is hard", 0),
]

def encode_fixed(text, max_len=4):
    ids = encode(text)
    ids = ids[:max_len]
    ids += [0] * (max_len - len(ids))
    return ids

ft_X = torch.tensor([encode_fixed(text, seq_len) for text, label in fine_tune_data])
ft_y = torch.tensor([label for text, label in fine_tune_data])
```

```python
class GPTFineTuner(nn.Module):
    def __init__(self, pretrained_gpt, d_model=32, num_classes=2):
        super().__init__()

        self.gpt = pretrained_gpt
        self.classifier = nn.Linear(d_model, num_classes)

    def forward(self, input_ids):
        batch_size, seq_len = input_ids.shape

        positions = torch.arange(seq_len).unsqueeze(0).expand(batch_size, seq_len)

        x = self.gpt.token_embedding(input_ids) + self.gpt.position_embedding(positions)

        mask = self.gpt.causal_mask(seq_len)

        x = self.gpt.transformer(x, mask=mask)

        final_hidden = x[:, -1, :]

        logits = self.classifier(final_hidden)

        return logits
```

```python
fine_model = GPTFineTuner(model)

optimizer = torch.optim.Adam(fine_model.parameters(), lr=0.005)
loss_fn = nn.CrossEntropyLoss()

for epoch in range(200):
    logits = fine_model(ft_X)
    loss = loss_fn(logits, ft_y)

    optimizer.zero_grad()
    loss.backward()
    optimizer.step()

    if (epoch + 1) % 50 == 0:
        pred = logits.argmax(dim=1)
        acc = (pred == ft_y).float().mean().item()
        print(f"Fine-tune Epoch {epoch+1}, Loss: {loss.item():.4f}, Acc: {acc:.2f}")
```

```python
def predict_sentiment(text):
    x = torch.tensor([encode_fixed(text, seq_len)])

    with torch.no_grad():
        logits = fine_model(x)

    pred = logits.argmax(dim=1).item()

    return "positive" if pred == 1 else "negative"

print(predict_sentiment("i love deep learning"))
print(predict_sentiment("machine learning is hard"))
```
