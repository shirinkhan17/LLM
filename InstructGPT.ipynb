
```text

```

# Remove Final Output Layer

They remove the normal GPT language-generation output layer:

```text

```

and replace it with:

```text

```

---

# New Goal

Instead of generating text, the reward model learns:

```text

→ reward score
```

Example:

| Response | Reward |
|---|---|
| helpful answer | 0.95 |
| bad answer | 0.10 |

# Human Comparison Dataset

For ONE prompt:

```text
Explain gravity simply.
```

GPT generates several answers:

| Output | Example |
|---|---|
| A | technical |
| B | simple |
| C | very clear |

Humans rank them:

```text
C > B > A
```

meaning:
- C preferred over B
- B preferred over A

---

# Training Objective

The reward model learns:

```text
preferred answers should get higher reward
```

---

# Example

Suppose:

```text

reward(B) = 0.6
```

Good.

But if:

```text
reward(C) = 0.2
reward(B) = 0.8
```

that is wrong,
so loss becomes large.

# Step 2 — Train Reward Model from Human Ranking

````python
import torch
import torch.nn as nn
from transformers import AutoTokenizer, AutoModel

MODEL_NAME = "gpt2"

tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)
tokenizer.pad_token = tokenizer.eos_token

# -------------------------------------------------
# Human ranked data
# Best answer first, worst answer last
# -------------------------------------------------

ranked_data = [
    {
        "prompt": "Explain gravity simply.",
        "responses": [
            "Gravity is a force that pulls objects toward each other.",
            "Gravity pulls things down toward the Earth.",
            "Gravity is a complicated physical interaction."
        ]
    },
    {
        "prompt": "Explain diabetes simply.",
        "responses": [
            "Diabetes is a condition where the body has trouble controlling blood sugar.",
            "Diabetes means blood sugar is too high.",
            "Diabetes is a metabolic disorder involving multiple pathways."
        ]
    }
]

# -------------------------------------------------
# Convert rankings into chosen/rejected pairs
# -------------------------------------------------

preference_data = []

for item in ranked_data:
    prompt = item["prompt"]
    responses = item["responses"]

    for i in range(len(responses)):
        for j in range(i + 1, len(responses)):
            preference_data.append({
                "prompt": prompt,
                "chosen": responses[i],
                "rejected": responses[j]
            })

# -------------------------------------------------
# Reward Model
# -------------------------------------------------

class RewardModel(nn.Module):
    def __init__(self):
        super().__init__()
        self.encoder = AutoModel.from_pretrained(MODEL_NAME)
        self.reward_head = nn.Linear(
            self.encoder.config.hidden_size,
            1
        )

    def forward(self, input_ids, attention_mask):
        outputs = self.encoder(
            input_ids=input_ids,
            attention_mask=attention_mask
        )

        hidden = outputs.last_hidden_state

        last_index = attention_mask.sum(dim=1) - 1
        batch_index = torch.arange(hidden.size(0))

        final_hidden = hidden[batch_index, last_index]

        reward = self.reward_head(final_hidden)

        return reward.squeeze(-1)

reward_model = RewardModel()

optimizer = torch.optim.Adam(
    reward_model.parameters(),
    lr=1e-5
)

# -------------------------------------------------
# Train Reward Model
# -------------------------------------------------

for epoch in range(5):
    total_loss = 0

    for item in preference_data:
        chosen_text = item["prompt"] + " " + item["chosen"]
        rejected_text = item["prompt"] + " " + item["rejected"]

        chosen_batch = tokenizer(
            chosen_text,
            return_tensors="pt",
            padding=True,
            truncation=True
        )

        rejected_batch = tokenizer(
            rejected_text,
            return_tensors="pt",
            padding=True,
            truncation=True
        )

        chosen_reward = reward_model(
            chosen_batch["input_ids"],
            chosen_batch["attention_mask"]
        )

        rejected_reward = reward_model(
            rejected_batch["input_ids"],
            rejected_batch["attention_mask"]
        )

        loss = -torch.log(
            torch.sigmoid(chosen_reward - rejected_reward)
        ).mean()

        optimizer.zero_grad()
        loss.backward()
        optimizer.step()

        total_loss += loss.item()

    print(f"Reward Model Epoch {epoch+1}, Loss={total_loss:.4f}")

# -------------------------------------------------
# Test Reward Model
# -------------------------------------------------

test_prompt = "Explain gravity simply."

answers = [
    "Gravity pulls objects toward each other.",
    "Gravity is a tensor equation in spacetime geometry."
]

for ans in answers:
    text = test_prompt + " " + ans

    batch = tokenizer(
        text,
        return_tensors="pt",
        padding=True,
        truncation=True
    )

    with torch.no_grad():
        score = reward_model(
            batch["input_ids"],
            batch["attention_mask"]
        )

    print("\nAnswer:", ans)
    print("Reward score:", score.item())
````

# Step 3 — Use Reward Model to Fine-Tune GPT with PPO

````python
ppo_model = AutoModelForCausalLMWithValueHead.from_pretrained(MODEL_NAME)
ref_model = AutoModelForCausalLMWithValueHead.from_pretrained(MODEL_NAME)

ppo_config = PPOConfig(
    learning_rate=1e-5,
    batch_size=1
)

ppo_trainer = PPOTrainer(
    config=ppo_config,
    model=ppo_model,
    ref_model=ref_model,
    tokenizer=tokenizer
)



prompts = [
    "Explain gravity simply.",
    "Explain diabetes simply."
]

reward_model.eval()

for prompt in prompts:
    query_ids = tokenizer(
        prompt,
        return_tensors="pt"
    )["input_ids"][0]

    response_ids = ppo_trainer.generate(
        query_ids,
        max_new_tokens=30,
        do_sample=True
    )

    response_text = tokenizer.decode(
        response_ids,
        skip_special_tokens=True
    )

    reward_batch = tokenizer(
        response_text,
        return_tensors="pt",
        padding=True,
        truncation=True
    )

    with torch.no_grad():
        reward_score = reward_model(
            reward_batch["input_ids"],
            reward_batch["attention_mask"]
        )

    ppo_trainer.step(
        [query_ids],
        [response_ids],
        [reward_score]
    )

    print("Prompt:", prompt)
    print("Response:", response_text)
    print("Reward:", reward_score.item())
    print()
````
