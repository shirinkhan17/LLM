# Claude 2 Architecture

Claude 2 is a large language model developed by Anthropic.

Exact architecture details are not fully public, but at a high level Claude 2 is a:

```text
Transformer-based autoregressive language model
```

---

# 1. Input Prompt

User prompt enters the model.

Example:

```text
Classify this forest cover type.
```

---

# 2. Tokenization

The text is converted into tokens.

```text
Text → Tokens
```

---

# 3. Embedding Layer

Tokens are converted into dense vector embeddings.

```text
Tokens → Embeddings
```

---

# 4. Transformer Layers

Claude 2 contains many Transformer layers.

Each layer includes:
- self-attention
- feedforward neural network
- normalization
- residual connections

---

# 5. Causal Self-Attention

Claude 2 is autoregressive.

Each token can attend only to:
- previous tokens
- current context

not future tokens.

```text
Previous Tokens
        ↓
Predict Next Token
```

Self-attention helps the model understand:
- instructions
- examples
- corrections
- conversation history

inside the context window.

---

# 6. Feedforward Network (FFN)

After attention, token representations pass through neural networks.

This helps learn:
- reasoning patterns
- semantic relationships
- language structure

---

# 7. Long Context Window

Claude 2 supports large context windows.

This allows it to process:
- long documents
- datasets
- multiple examples
- previous corrections
- long conversations

inside one prompt context.

---

# 8. Output Generation

Claude predicts the next token repeatedly.

```text
Prompt
   ↓
Tokenizer
   ↓
Embeddings
   ↓
Transformer Layers
   ↓
Next-token Probabilities
   ↓
Generated Response
```

---

# 9. Training Pipeline

Claude 2 is trained using:

```text
Pretraining
    ↓
Supervised Fine-Tuning (SFT)
    ↓
RLHF / Constitutional AI
    ↓
Aligned Assistant Model
```

---

# 10. Constitutional AI

Anthropic introduced Constitutional AI.

The model learns principles such as:
- helpfulness
- harmlessness
- honesty
- safe behavior

This improves:
- alignment
- safety
- conversational quality

---

# Full Claude 2 Flow

```text
User Prompt
      ↓
Tokenizer
      ↓
Token Embeddings
      ↓
Transformer Layers
      ↓
Causal Self-Attention
      ↓
Feedforward Networks
      ↓
Next-token Prediction
      ↓
Generated Response
```

---

# Important

Claude 2 is NOT rule-based.

It is a pretrained neural Transformer model.

During prompting:
- weights do NOT change
- no retraining occurs

The model temporarily adapts behavior using:
- prompt context
- examples
- corrections
- self-attention over conversation history

# Batch Learning vs Online Learning
![My Image](image/7.png)
## Batch Learning

Batch learning trains the model using the entire dataset at once.

Pipeline:

```text
Full Dataset
      ↓
Train/Test Split
      ↓
Train Model
      ↓
Fixed Trained Model
      ↓
Prediction
```

Example:
- Decision Tree
- Random Forest
- SVM

The model:
- sees all training data together
- trains once
- parameters are updated during training
- remains fixed after training

If new data arrives:
- retraining is usually required.

---

# Example of Batch Learning

```text
Dataset:
10000 labeled forest samples
```

The model trains on:
- all 10000 samples together

After training:

```text
Model weights/parameters become fixed
```

Then prediction starts.

---

# Online Learning

Online learning processes data incrementally as a stream.

Pipeline:

```text
Data Stream
     ↓
Small Batch Arrives
     ↓
Update Model
     ↓
Next Batch Arrives
     ↓
Update Model Again
```

Example:
- Hoeffding Tree
- Online Naive Bayes
- Adaptive Random Forest

The model:
- learns continuously
- updates parameters repeatedly
- adapts to new incoming data

---

# Example of Online Learning

```text
Day 1 data arrives
    ↓
Model update

Day 2 data arrives
    ↓
Another update

Day 3 data arrives
    ↓
Another update
```

The model keeps changing over time.

---

# Main Difference

| Batch Learning | Online Learning |
|---|---|
| trains once | trains continuously |
| uses full dataset | uses streaming batches |
| fixed after training | continuously updated |
| retraining needed for new data | adapts automatically |
| static learning | incremental learning |

---

# Important

In online learning:

```text
model parameters actually change continuously
```

through:
- gradient updates
- incremental optimization

Unlike prompt-based LLM adaptation,
online learning performs real parameter updates.

A pretrained LLM like Claude or GPT-4 has already learned:
- language
- instructions
- reasoning patterns
- examples
- corrections

during massive pretraining.

During prompting, the model uses the prompt context to guide output generation.

```text
Prompt
   ↓
Tokenizer
   ↓
Transformer Attention
   ↓
Understand context/instructions
   ↓
Shift token probabilities
   ↓
Generate output
```

When the model receives many instructions/examples, the Transformer processes all of them together through self-attention.

```text
Instructions/examples
        +
Current query
        ↓
Transformer Attention
        ↓
Context-conditioned reasoning
        ↓
Output generation
```

The output is conditioned on:
- instructions
- examples
- corrections
- retrieved documents
- previous conversation

inside the context window.

---

Suppose the model receives:

```text
You are a biomedical extractor.

Example:
KMT2B → dystonia

Now analyze:
TAF1 mutations cause dystonia.
```

The model may incorrectly output:

```json
{
  "gene": "TAF1",
  "disease": "parkinsonism"
}
```

Now the user sends a correction:

```text
The previous prediction was incorrect.

For:
"TAF1 mutations cause dystonia"

the correct disease is:
dystonia
```

The model is NOT retrained:
- weights do NOT change
- no backpropagation occurs

Instead, the correction becomes part of the active context.

Now the context contains:
- original example
- wrong prediction
- correction text

When the next query arrives:

```text
Now analyze:
ABC1 mutations cause ataxia.
```

the Transformer attention mechanism can look back at:
- previous examples
- correction text
- instructions

and infer the corrected extraction pattern.

The model may now output:

```json
{
  "gene": "ABC1",
  "disease": "ataxia"
}
```

So the correction works because:
- the correction is added into the context
- self-attention can access earlier conversation text
- future outputs become conditioned on that correction

This is:
- temporary context adaptation
- NOT permanent learning

# Prompt Engineering Techniques

| Technique | Purpose |
|---|---|
| Zero-shot prompting | instruction only |
| Few-shot prompting | examples inside prompt |
| Chain-of-thought | step-by-step reasoning |
| RAG | retrieved documents added to prompt |
| Agent prompting | tool-using multi-step reasoning |

---

# Difference Between Online Learning and Prompt Engineering

| Online Learning | Prompt Engineering |
|---|---|
| updates model parameters | updates context only |
| real training | no training |
| permanent learning | temporary adaptation |
| gradient updates | attention-based reasoning |

---

# Important

LLMs like:
- Claude

can simulate online-like behavior using prompts and corrections,
but the model weights remain unchanged unless actual fine-tuning occurs.

````python
# Simulate Claude-style online prompting with feedback
# No model weights are updated.
# The "learning" happens by adding feedback into the conversation context.

class MockLLM:
    def __init__(self):
        self.memory = []

    def predict(self, batch):
        predictions = []

        for item in batch:
            elevation = item["elevation"]
            soil = item["soil"]
            rain = item["rain"]

            # Simple rule-based mock behavior
            # This imitates an LLM making predictions from context
            if soil == "Dry" and rain == "Low":
                pred = "Pine Forest"
            elif soil == "Rocky" and rain == "High":
                # After feedback, model adapts its rule
                if "Rocky high elevation can indicate Fir Forest" in self.memory:
                    pred = "Fir Forest"
                else:
                    pred = "Spruce Forest"
            else:
                pred = "Unknown"

            predictions.append(pred)

        return predictions

    def receive_feedback(self, feedback):
        self.memory.append(feedback)


# -----------------------------
# Initial labeled examples
# -----------------------------

initial_examples = [
    {
        "elevation": 3000,
        "soil": "Rocky",
        "rain": "High",
        "label": "Spruce Forest"
    },
    {
        "elevation": 1500,
        "soil": "Dry",
        "rain": "Low",
        "label": "Pine Forest"
    }
]

print("Initial examples given to LLM:")
for ex in initial_examples:
    print(ex)


# -----------------------------
# Test batch 1
# -----------------------------

batch_1 = [
    {"elevation": 1600, "soil": "Dry", "rain": "Low"},
    {"elevation": 2900, "soil": "Rocky", "rain": "High"}
]

llm = MockLLM()

pred_1 = llm.predict(batch_1)

print("\nPredictions for batch 1:")
for item, pred in zip(batch_1, pred_1):
    print(item, "→", pred)


# -----------------------------
# Correct answers given back
# -----------------------------

correct_answers_1 = ["Pine Forest", "Fir Forest"]

print("\nCorrect answers for batch 1:")
for item, pred, gold in zip(batch_1, pred_1, correct_answers_1):
    print(item)
    print("Predicted:", pred)
    print("Correct:", gold)

    if pred != gold:
        feedback = "Rocky high elevation can indicate Fir Forest"
        llm.receive_feedback(feedback)
        print("Feedback added:", feedback)


# -----------------------------
# Test batch 2
# -----------------------------

batch_2 = [
    {"elevation": 3100, "soil": "Rocky", "rain": "High"},
    {"elevation": 1700, "soil": "Dry", "rain": "Low"}
]

pred_2 = llm.predict(batch_2)

print("\nPredictions for batch 2 after feedback:")
for item, pred in zip(batch_2, pred_2):
    print(item, "→", pred)
````
