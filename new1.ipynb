```python
import streamlit as st
import openai

# Set your OpenAI API key
openai.api_key = "YOUR_API_KEY"

st.title("🧬 Phenotype-to-Disease Matcher")
st.markdown("Enter clinical features or HPO terms to get matching rare diseases.")

# User input
symptoms = st.text_area("🩺 Enter symptoms or HPO terms:",
                       placeholder="e.g. hypotonia, seizures, developmental delay")

# Button to trigger LLM
if st.button("🔍 Find Matching Diseases"):
    if symptoms.strip() == "":
        st.warning("Please enter at least one symptom.")
    else:
        with st.spinner("Analyzing with GPT-4..."):
            prompt = f"""
The patient presents with the following symptoms:
{symptoms}

List the most likely rare diseases. For each, provide:
1. Disease name
2. Associated gene(s)
3. Inheritance pattern
4. Brief explanation of the match

Limit to 3-5 likely diseases.
"""
            
            try:
                response = openai.ChatCompletion.create(
                    model="gpt-4",
                    messages=[
                        {"role": "system", "content": "You are a rare disease expert."},
                        {"role": "user", "content": prompt}
                    ],
                    temperature=0.7,
                    max_tokens=800
                )
                answer = response['choices'][0]['message']['content']
                st.markdown("### 🧾 Possible Matches:")
                st.markdown(answer)
            except Exception as e:
                st.error(f"Error: {str(e)}")
```
