# T5: Exploring the Limits of Transfer Learning with a Unified Text-to-Text Transformer

Paper:
- Google Research (2020)
- Introduced:
  
```text
T5 = Text-To-Text Transfer Transformer
```

Main idea:

```text
Convert EVERY NLP task into a text-to-text problem.
```

---

# Core Idea

Before T5:
- different NLP tasks often used:
  - different architectures
  - different output layers
  - different training pipelines

T5 proposed:

```text
ONE unified framework
```

where:
- input = text
- output = text

for ALL tasks.

---

# Main Philosophy

T5 treats every task as:

```text
text → text
```

---

# Examples

## Translation

Input:

```text
translate English to French: cat
```

Output:

```text
chat
```

---

# Summarization

Input:

```text
summarize: The patient was diagnosed with diabetes...
```

Output:

```text
patient received diabetes treatment
```

---

# Question Answering

Input:

```text
question: What color is the sky?
context: The sky is blue.
```

Output:

```text
blue
```

---

# Sentiment Analysis

Input:

```text

```

Output:

```text
positive
```

---

# Key Innovation

Instead of:
- classifiers
- separate heads
- task-specific architectures

T5 uses:

```text
one encoder-decoder Transformer
```

for EVERYTHING.



# Why Encoder-Decoder?

Because many NLP tasks require BOTH:
- understanding input
- generating output

Encoder:
- understands text bidirectionally

Decoder:
- generates output autoregressively

---


instead of BERT MLM.

Model learns to reconstruct missing spans.

# Architecture Difference:
# T5 vs Attention Is All You Need

Actually:

```text
T5 uses almost the SAME base Transformer architecture
introduced in Attention Is All You Need.
```

BUT T5 modifies and extends it.

---

# 1. Attention Is All You Need Architecture

Original Transformer:

```text
Encoder
   ↓
Decoder
```

Each encoder layer:
- self-attention
- feedforward network

Each decoder layer:
- masked self-attention
- encoder-decoder attention
- feedforward network

---

# Original Transformer Flow

```text
Input Sentence
      ↓
Encoder
      ↓
Context Representation
      ↓
Decoder
      ↓
Generated Output
```

Main purpose:
- translation

---

# 2. T5 Architecture

T5 ALSO uses:

```text
encoder-decoder Transformer
```

BUT with modifications.

---

# Major Architectural Differences

| Feature | Attention Is All You Need | T5 |
|---|---|---|
| Position Encoding | sinusoidal positional encoding | learned relative position bias |
| LayerNorm placement | after sublayer | before sublayer |
| Objective | supervised translation | span corruption pretraining |
| Task Design | translation-focused | unified text-to-text |
| Pretraining | no | yes |
| Transfer Learning | limited | central idea |

---

# 3. Positional Encoding Difference

## Attention Is All You Need

Uses:

```text
fixed sinusoidal positional encoding
```

Example:

\[
PE(pos,2i)=\sin(pos/10000^{2i/d})
\]

Positions are:
- mathematical
- fixed
- not learned

---

# T5

Uses:

```text
relative position embeddings/bias
```

Meaning:
- model learns relative distances between tokens

Example:
- token is 2 words away
- token is 5 words away

instead of absolute positions.

---

# 4. LayerNorm Difference

## Original Transformer

```text
Attention
   ↓
Add & Norm
```

LayerNorm AFTER residual connection.

---

# T5

Uses:

```text
Norm
  ↓
Attention
```

LayerNorm BEFORE attention/feedforward.

Called:

```text
Pre-LN Transformer
```

This improves:
- training stability
- large-scale training

---

# 5. Pretraining Difference

## Attention Is All You Need

Mostly:

```text
supervised translation training
```

No large language pretraining.

---

# T5

Massive:
- self-supervised pretraining
- transfer learning

---

# 6. Objective Difference

## Original Transformer

```text

```

---

## T5

Example:

## Attention Is All You Need

Mostly:
- one task

## T5

Everything becomes:

```text

```

using SAME architecture.

# Positional Embedding Difference:
# T5 vs Attention Is All You Need

The main difference is:

| Model | Position Method |
|---|---|
| Attention Is All You Need | absolute sinusoidal positional encoding |
| T5 | learned relative position bias |

---

# 1. Attention Is All You Need

Transformer itself has NO notion of order.

So original Transformer added:

```text
absolute positional encoding
```

to token embeddings.

---

# Example

Sentence:

```text
The cat sat on mat
```

Positions:

| Token | Position |
|---|---|
| The | 0 |
| cat | 1 |
| sat | 2 |
| on | 3 |
| mat | 4 |

---

# Original Transformer Encoding

Uses mathematical functions:

\[
PE(pos,2i)=\sin(pos/10000^{2i/d})
\]

\[
PE(pos,2i+1)=\cos(pos/10000^{2i/d})
\]

These are:

```text
fixed sinusoidal embeddings
```

---

# Important Properties

## Absolute Position

Each token gets:
- its own position vector

Example:

```text
Position 0 → one vector
Position 1 → another vector
```

---

# Fixed

They are:
- NOT learned
- manually computed

---

# Added to Token Embedding

```text
TokenEmbedding + PositionalEncoding
```

---

# 2. T5 Position Method

T5 does NOT use:
- sinusoidal encoding
- absolute positions

Instead T5 uses:

```text
relative position bias
```

inside attention.

---

# Main Idea

T5 cares about:

```text
distance between tokens
```

rather than exact position.

---

# Example

Sentence:

```text
The cat sat on mat
```

When token:

```text
sat
```

attends to:

```text
cat
```

T5 learns:

```text
distance = -1
```

When attending to:

```text
mat
```

distance:

```text
+2
```

---

# Relative Position

Attention learns:

```text
how far tokens are from each other
```

instead of:
- absolute index numbers

---

# Where It Is Applied

Original Transformer attention:

:contentReference[oaicite:0]{index=0}

T5 modifies attention scores by adding:

```text
relative position bias
```

before softmax.

---

# Why Relative Position Helps

Language often depends more on:

```text
relative distance
```

than absolute location.

Example:
- previous word
- nearby phrase
- neighboring tokens

matter more than:
- "this is token #57"

---

# Advantages of T5 Relative Position Bias

## Better Generalization

Works better for:
- longer sequences
- unseen lengths

---

# More Flexible

Learns:
- nearby token importance
- local/global relationships

more naturally.

---

# Learned

Unlike sinusoidal encoding:

```text
T5 learns position relationships
```

during training.

---

# Simple Comparison

## Attention Is All You Need

```text
"I am token at position 12"
```

---

## T5

```text
"I am 2 words away from this token"
```

---

# Visualization

## Original Transformer

```text
Token + Absolute Position
```

Example:

```text
cat + position_3
```

---

## T5

```text
Attention(token_i, token_j)
+ relative_distance_bias(i,j)
```

---

# Final Intuition

## Attention Is All You Need

Uses:

```text
absolute fixed positional encoding
```

to inject token order.

---

## T5

Uses:

```text
learned relative positional bias
```

inside attention to model token distances more flexibly.

````python
import torch
import torch.nn as nn
import torch.nn.functional as F

torch.manual_seed(42)

# -------------------------------------------------
# Tiny T5 Attention With Relative Position Bias
# -------------------------------------------------

class RelativeAttention(nn.Module):

    def __init__(self, d_model, num_heads, max_len=32):
        super().__init__()

        self.num_heads = num_heads
        self.head_dim = d_model // num_heads

        self.q = nn.Linear(d_model, d_model)
        self.k = nn.Linear(d_model, d_model)
        self.v = nn.Linear(d_model, d_model)

        self.out = nn.Linear(d_model, d_model)

        # T5-style relative position bias
        self.relative_bias = nn.Parameter(
            torch.randn(num_heads, max_len, max_len)
        )

    def forward(self, x):

        B, T, C = x.shape

        Q = self.q(x)
        K = self.k(x)
        V = self.v(x)

        # Split heads
        Q = Q.view(B, T, self.num_heads, self.head_dim).transpose(1, 2)
        K = K.view(B, T, self.num_heads, self.head_dim).transpose(1, 2)
        V = V.view(B, T, self.num_heads, self.head_dim).transpose(1, 2)

        # Attention scores
        scores = torch.matmul(
            Q,
            K.transpose(-2, -1)
        )

        scores = scores / (self.head_dim ** 0.5)

        # -------------------------------------------------
        # ADD RELATIVE POSITION BIAS HERE
        # -------------------------------------------------

        bias = self.relative_bias[:, :T, :T]

        scores = scores + bias.unsqueeze(0)

        # Attention probabilities
        attn = F.softmax(scores, dim=-1)

        # Attention output
        out = torch.matmul(attn, V)

        out = out.transpose(1, 2).contiguous()

        out = out.view(B, T, C)

        return self.out(out)


# -------------------------------------------------
# Tiny T5 Block
# -------------------------------------------------

class TinyT5Block(nn.Module):

    def __init__(self, d_model, num_heads):
        super().__init__()

        self.attn = RelativeAttention(
            d_model,
            num_heads
        )

        self.norm1 = nn.LayerNorm(d_model)

        self.ff = nn.Sequential(
            nn.Linear(d_model, 64),
            nn.ReLU(),
            nn.Linear(64, d_model)
        )

        self.norm2 = nn.LayerNorm(d_model)

    def forward(self, x):

        x = x + self.attn(
            self.norm1(x)
        )

        x = x + self.ff(
            self.norm2(x)
        )

        return x


# -------------------------------------------------
# Tiny T5
# -------------------------------------------------

class TinyT5(nn.Module):

    def __init__(
        self,
        vocab_size,
        d_model=32,
        num_heads=4,
        max_len=32
    ):
        super().__init__()

        # Shared embedding
        self.shared_embedding = nn.Embedding(
            vocab_size,
            d_model
        )

        self.encoder = TinyT5Block(
            d_model,
            num_heads
        )

        self.decoder = TinyT5Block(
            d_model,
            num_heads
        )

        self.lm_head = nn.Linear(
            d_model,
            vocab_size
        )

    def forward(
        self,
        encoder_input_ids,
        decoder_input_ids
    ):

        # Encoder
        enc = self.shared_embedding(
            encoder_input_ids
        )

        memory = self.encoder(enc)

        # Decoder
        dec = self.shared_embedding(
            decoder_input_ids
        )

        out = self.decoder(dec)

        logits = self.lm_head(out)

        return logits


# -------------------------------------------------
# Vocabulary
# -------------------------------------------------

vocab = {
    "<pad>": 0,
    "<bos>": 1,
    "patient": 2,
    "received": 3,
    "treatment": 4,
    "diabetes": 5,
}

id_to_word = {
    v: k for k, v in vocab.items()
}


# -------------------------------------------------
# Example
# -------------------------------------------------

encoder_input_ids = torch.tensor([
    [2, 5]
])

decoder_input_ids = torch.tensor([
    [1, 2]
])

target_ids = torch.tensor([
    [2, 3]
])


# -------------------------------------------------
# Model
# -------------------------------------------------

model = TinyT5(
    vocab_size=len(vocab)
)

optimizer = torch.optim.Adam(
    model.parameters(),
    lr=0.01
)

loss_fn = nn.CrossEntropyLoss()


# -------------------------------------------------
# Train
# -------------------------------------------------

for epoch in range(200):

    logits = model(
        encoder_input_ids,
        decoder_input_ids
    )

    loss = loss_fn(
        logits.view(-1, len(vocab)),
        target_ids.view(-1)
    )

    optimizer.zero_grad()

    loss.backward()

    optimizer.step()

    if (epoch + 1) % 50 == 0:
        print(
            f"Epoch {epoch+1}, Loss={loss.item():.4f}"
        )


# -------------------------------------------------
# Predict
# -------------------------------------------------

model.eval()

with torch.no_grad():

    logits = model(
        encoder_input_ids,
        decoder_input_ids
    )

prediction = logits.argmax(dim=-1)

print(prediction)

print("\nPredicted Tokens:")

for token_id in prediction[0]:
    print(
        id_to_word[token_id.item()],
        end=" "
    )
````
