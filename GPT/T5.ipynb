# T5: Exploring the Limits of Transfer Learning with a Unified Text-to-Text Transformer

- Google Research (2020)

```text
T5 = Text-To-Text Transfer Transformer
```

```text
Convert EVERY NLP task into a text-to-text problem.
```

---

  - different output layers
  - different training pipelines

```text
ONE unified framework
```

where:
- input = text
- output = text

for ALL tasks.

```text

```

# Architecture Difference:
# T5 vs Attention Is All You Need

Actually:

```text
T5 uses almost the SAME base Transformer architecture
introduced in Attention Is All You Need.
```

BUT T5 modifies and extends it.

---

# 1. Attention Is All You Need Architecture

```text

   ↓

```

- feedforward network

- masked self-attention

# Original Transformer Flow

```text

      ↓
Encoder
      ↓
Context Representation
      ↓
Decoder
      ↓
Generated Output
```

Main purpose:
- translation

---

# 2. T5 Architecture

T5 ALSO uses:

```text
encoder-decoder Transformer
```

BUT with modifications.

---

# Major Architectural Differences

| Feature | Attention Is All You Need | T5 |
|---|---|---|
| Position Encoding | sinusoidal positional encoding | learned relative position bias |
| LayerNorm placement | after sublayer | before sublayer |
| Objective | supervised translation | span corruption pretraining |
| Task Design | translation-focused | unified text-to-text |
| Pretraining | no | yes |
| Transfer Learning | limited | central idea |

---

# 3. Positional Encoding Difference

## Attention Is All You Need

Uses:

```text
fixed sinusoidal positional encoding
```

Example:

\[
PE(pos,2i)=\sin(pos/10000^{2i/d})
\]

Positions are:
- mathematical
- fixed
- not learned

---

# T5

Uses:

```text
relative position embeddings/bias
```

Meaning:
- model learns relative distances between tokens

Example:
- token is 2 words away
- token is 5 words away

instead of absolute positions.

---

# 4. LayerNorm Difference

## Original Transformer

```text
Attention
   ↓
Add & Norm
```

LayerNorm AFTER residual connection.

---

# T5

Uses:

```text
Norm
  ↓
Attention
```

LayerNorm BEFORE attention/feedforward.

Called:

```text
Pre-LN Transformer
```

This improves:
- training stability
- large-scale training

# Positional Embedding Difference:
# T5 vs Attention Is All You Need

The main difference is:

| Model | Position Method |
|---|---|
| Attention Is All You Need | absolute sinusoidal positional encoding |
| T5 | learned relative position bias |

---

# 1. Attention Is All You Need

Transformer itself has NO notion of order.

So original Transformer added:

```text
absolute positional encoding
```

to token embeddings.

---

# Example

Sentence:

```text
The cat sat on mat
```

Positions:

| Token | Position |
|---|---|
| The | 0 |
| cat | 1 |
| sat | 2 |
| on | 3 |
| mat | 4 |

---

# Original Transformer Encoding

Uses mathematical functions:

\[
PE(pos,2i)=\sin(pos/10000^{2i/d})
\]

\[
PE(pos,2i+1)=\cos(pos/10000^{2i/d})
\]

These are:

```text
fixed sinusoidal embeddings
```

---

# Important Properties

## Absolute Position

Each token gets:
- its own position vector

Example:

```text
Position 0 → one vector
Position 1 → another vector
```

---

# Fixed

They are:
- NOT learned
- manually computed

---

# Added to Token Embedding

```text
TokenEmbedding + PositionalEncoding
```

---

# 2. T5 Position Method

T5 does NOT use:
- sinusoidal encoding
- absolute positions

Instead T5 uses:

```text
relative position bias
```

inside attention.

---

# Main Idea

T5 cares about:

```text
distance between tokens
```

rather than exact position.

---

# Example

Sentence:

```text
The cat sat on mat
```

When token:

```text
sat
```

attends to:

```text
cat
```

T5 learns:

```text
distance = -1
```

When attending to:

```text
mat
```

distance:

```text
+2
```

---

# Relative Position

Attention learns:

```text
how far tokens are from each other
```

instead of:
- absolute index numbers

---

# Where It Is Applied

Original Transformer attention:

:contentReference[oaicite:0]{index=0}

T5 modifies attention scores by adding:

```text
relative position bias
```

before softmax.

---

# Why Relative Position Helps

Language often depends more on:

```text
relative distance
```

than absolute location.

Example:
- previous word
- nearby phrase
- neighboring tokens

matter more than:
- "this is token #57"

---

# Advantages of T5 Relative Position Bias

## Better Generalization

Works better for:
- longer sequences
- unseen lengths

---

# More Flexible

Learns:
- nearby token importance
- local/global relationships

more naturally.

---

# Learned

Unlike sinusoidal encoding:

```text
T5 learns position relationships
```

during training.

---


---

## T5



````python
import torch
import torch.nn as nn
import random

torch.manual_seed(42)
random.seed(42)

# =====================================================
# 1. Raw unlabeled text
# =====================================================

raw_texts = [
    "patient received medicine at hospital",
    "doctor treated patient at hospital",
    "bank approved loan for customer",
    "market affected money investment"
]

# =====================================================
# 2. Vocabulary
# =====================================================

special_tokens = ["<pad>", "<bos>", "<eos>", "<extra_id_0>"]

all_words = set(special_tokens)

for text in raw_texts:
    all_words.update(text.split())

vocab = {word: i for i, word in enumerate(sorted(all_words))}
id_to_word = {i: w for w, i in vocab.items()}

PAD = vocab["<pad>"]
BOS = vocab["<bos>"]
EOS = vocab["<eos>"]
MASK = vocab["<extra_id_0>"]

# =====================================================
# 3. Automatic T5-style span corruption
# =====================================================

def make_t5_example(text):
    tokens = text.split()

    # automatically mask middle span
    start = 1
    end = min(3, len(tokens))

    masked_span = tokens[start:end]

    encoder_tokens = (
        tokens[:start]
        + ["<extra_id_0>"]
        + tokens[end:]
        + ["<eos>"]
    )

    target_tokens = (
        ["<extra_id_0>"]
        + masked_span
        + ["<eos>"]
    )

    return encoder_tokens, target_tokens


def encode(tokens, max_len):
    ids = [vocab[t] for t in tokens]
    ids = ids[:max_len]
    ids += [PAD] * (max_len - len(ids))
    return ids


max_encoder_len = 6
max_decoder_len = 5

encoder_inputs = []
decoder_inputs = []
targets = []

for text in raw_texts:
    encoder_tokens, target_tokens = make_t5_example(text)

    decoder_tokens = ["<bos>"] + target_tokens[:-1]

    encoder_inputs.append(encode(encoder_tokens, max_encoder_len))
    decoder_inputs.append(encode(decoder_tokens, max_decoder_len))
    targets.append(encode(target_tokens, max_decoder_len))

encoder_input_ids = torch.tensor(encoder_inputs)
decoder_input_ids = torch.tensor(decoder_inputs)
target_ids = torch.tensor(targets)

# =====================================================
# 4. Tiny T5 model
# =====================================================

class TinyT5(nn.Module):
    def __init__(self, vocab_size, d_model=32, num_heads=4, num_layers=2):
        super().__init__()

        self.embedding = nn.Embedding(vocab_size, d_model)

        self.transformer = nn.Transformer(
            d_model=d_model,
            nhead=num_heads,
            num_encoder_layers=num_layers,
            num_decoder_layers=num_layers,
            dim_feedforward=64,
            batch_first=True
        )

        self.lm_head = nn.Linear(d_model, vocab_size)

    def causal_mask(self, size):
        return torch.triu(torch.ones(size, size), diagonal=1).bool()

    def forward(self, encoder_input_ids, decoder_input_ids):
        enc = self.embedding(encoder_input_ids)
        dec = self.embedding(decoder_input_ids)

        tgt_len = decoder_input_ids.size(1)
        tgt_mask = self.causal_mask(tgt_len)

        output = self.transformer(
            src=enc,
            tgt=dec,
            tgt_mask=tgt_mask
        )

        logits = self.lm_head(output)

        return logits

# =====================================================
# 5. Train self-supervised T5-style model
# =====================================================

model = TinyT5(vocab_size=len(vocab))

optimizer = torch.optim.Adam(model.parameters(), lr=0.01)
loss_fn = nn.CrossEntropyLoss(ignore_index=PAD)

for epoch in range(300):
    logits = model(encoder_input_ids, decoder_input_ids)

    loss = loss_fn(
        logits.reshape(-1, len(vocab)),
        target_ids.reshape(-1)
    )

    optimizer.zero_grad()
    loss.backward()
    optimizer.step()

    if (epoch + 1) % 50 == 0:
        print(f"Epoch {epoch+1}, Loss={loss.item():.4f}")

# =====================================================
# 6. Generate missing span
# =====================================================

def generate(text, max_len=5):
    model.eval()

    encoder_tokens, target_tokens = make_t5_example(text)

    encoder_ids = torch.tensor([
        encode(encoder_tokens, max_encoder_len)
    ])

    generated = [BOS]

    for _ in range(max_len):
        decoder_ids = torch.tensor([generated])

        with torch.no_grad():
            logits = model(encoder_ids, decoder_ids)

        next_id = logits[:, -1, :].argmax(dim=-1).item()

        if next_id == EOS:
            break

        generated.append(next_id)

    words = [
        id_to_word[i]
        for i in generated
        if i not in [BOS, EOS, PAD]
    ]

    return encoder_tokens, target_tokens, words

# =====================================================
# 7. Test
# =====================================================

test_text = "patient received medicine at hospital"

encoder_tokens, target_tokens, predicted_tokens = generate(test_text)

print("\nOriginal Text:")
print(test_text)

print("\nCorrupted Encoder Input:")
print(" ".join(encoder_tokens))

print("\nTarget Created Automatically:")
print(" ".join(target_tokens))

print("\nModel Prediction:")
print(" ".join(predicted_tokens))
````
