# GPT-2 vs GPT-3 Architecture Difference

Surprisingly:

```text
GPT-2 and GPT-3 use almost the SAME architecture.
```

Both are:

```text
decoder-only Transformer
```

using:
- causal self-attention
- autoregressive next-token prediction

---

# Core Architecture

Both use:

```text
Input Tokens
      ↓
Token Embeddings
      ↓
Masked Self-Attention
      ↓
Feedforward Layers
      ↓
Next Token Prediction
```

---

# Main Difference

The BIGGEST difference is:

```text
SCALE
```

NOT a fundamentally new architecture.

---

# Architecture Comparison

| Feature | GPT-2 | GPT-3 |
|---|---|
| Transformer type | decoder-only | decoder-only |
| Attention | causal self-attention | causal self-attention |
| Positional embeddings | learned absolute | learned absolute |
| Objective | next-token prediction | next-token prediction |
| Feedforward blocks | yes | yes |
| Residual connections | yes | yes |
| LayerNorm | yes | yes |

Very similar architecture.

---

# Major Scaling Differences

| Component | GPT-2 | GPT-3 |
|---|---|
| Parameters | 1.5B | 175B |
| Layers | 48 | 96 |
| Hidden size | 1600 | 12288 |
| Attention heads | 25 | 96 |
| Context length | 1024 | 2048 |
| Training data | smaller | much larger |

# Why GPT-3 Has 175B Parameters but GPT-2 Has 1.5B

The parameter count grows because GPT-3 is made MUCH larger in several dimensions simultaneously.

Transformer parameters mainly come from:

```text
- embedding matrices
- attention layers
- feedforward layers
```

---

# Main Scaling Components

| Component | GPT-2 | GPT-3 |
|---|---|---|
| layers | 48 | 96 |

| attention heads | 25 | 96 |

When these increase, parameter count explodes.

This becomes HUGE when `d_model` grows.

# GPT-2 vs GPT-3 Context Window
# 1024 vs 2048 Tokens

| Model | Context Window |
|---|---|
| GPT-2 | 1024 tokens |
| GPT-3 | 2048 tokens |

---

# What Is Context Window?

Context window means:

```text
how many previous tokens the model can see at once
```

during attention.

---

# Example

Sentence:

```text
The patient received medicine at the hospital.
```

Each word/subword becomes tokens.

Transformer attends over:

```text
previous context tokens
```

---

# GPT-2

GPT-2 can only look at:

```text
last 1024 tokens
```

Maximum.

Anything older:
- disappears from attention
- cannot be directly used.

---

# GPT-3

GPT-3 can look at:

```text
last 2048 tokens
```

So:
- longer documents
- bigger prompts
- more examples

fit inside model memory.

---

# Why Important for Few-Shot Learning

Few-shot prompting needs:

```text
multiple examples inside prompt
```

Example:

```text
English: cat
French: chat

English: dog
French: chien

English: house
French:
```

Larger context window means:
- more demonstrations
- more instructions
- longer reasoning chains

---

Context window also determines:

```text
maximum positional embeddings
```

GPT models use:

```text
learned absolute positional embeddings
```

```text
0 → 1023
```

GPT-3 learns:

```text
0 → 2047
```

```text
maximum sequence length visible during one forward pass
```
