# Language Models are Unsupervised Multitask Learners (GPT-2 Paper)



Paper:
- OpenAI (2019)
- Introduced GPT-2

Main idea:

```text
A sufficiently large language model trained only on next-word prediction
can perform many tasks without task-specific supervised training.
```

This was a major breakthrough.

---

# Core Idea

Before GPT-2:
- NLP systems usually needed:
  - supervised datasets
  - task-specific architectures
  - separate fine-tuning

GPT-2 showed:

```text
one large language model
trained only on language modeling
```

can perform MANY tasks simply from prompts.

---

# Training Objective

GPT-2 uses:

```text
autoregressive language modeling
```

Meaning:

predict next token.

Example:

```text
"The cat sat on the"
```

predict:

```text
mat
```

Training objective:

\[
P(x_t \mid x_1, x_2, ..., x_{t-1})
\]

---

# Architecture

GPT-2 uses:
- Transformer decoder
- causal self-attention
- left-to-right generation

NOT:
- encoder-decoder
- bidirectional encoder

---

# Key Discovery

When trained on:
- huge internet text
- very large scale

GPT-2 learned:
- translation
- summarization
- QA
- reasoning
- text completion

WITHOUT explicit supervised training.

---

# Why "Multitask"?

Because one model could do many tasks.

Example prompt:

## Translation

```text
English: Hello
French:
```

GPT-2 generates:

```text
Bonjour
```

---

## Summarization

```text
Article:
...
Summary:
```

GPT-2 generates summary.

---

## QA

```text
Question: What color is the sky?
Answer:
```

GPT-2 generates:

```text
blue
```

---

# Important Insight

Tasks can be expressed as:

```text
text prediction problems
```

instead of:
- separate architectures
- special loss functions

---

# Zero-Shot Learning

GPT-2 demonstrated:

```text
zero-shot task performance
```

Meaning:
- no fine-tuning
- no task-specific training

Only prompting.

---

# Example

Prompt:

```text
Translate English to French:

dog → chien
cat → chat
house →
```

GPT-2 infers task pattern and predicts:

```text
maison
```

---

# Why It Was Important

Before this paper:
- people believed supervised fine-tuning was necessary

GPT-2 showed:
- scale alone creates general abilities

This influenced:
- GPT-3
- ChatGPT
- modern LLMs

---

# Scaling Hypothesis

Paper strongly suggested:

```text
larger models + more data
→ emergent capabilities
```

---

# Training Data

GPT-2 trained on:
- WebText dataset
- internet pages from Reddit links

Much larger and more diverse than earlier corpora.

---

# Model Sizes

| Model | Parameters |
|---|---|
| Small | 117M |
| Medium | 345M |
| Large | 774M |
| XL | 1.5B |

1.5B was huge at the time.

---

# Why GPT-2 Worked So Well

Large-scale next-token prediction forces model to learn:
- grammar
- facts
- reasoning patterns
- semantic relationships
- world knowledge

because predicting language requires understanding language.

---

# Key Difference from BERT

| GPT-2 | BERT |
|---|---|
| decoder-only | encoder-only |
| autoregressive | masked LM |
| generation-focused | understanding-focused |
| causal attention | bidirectional attention |

---

# Big Picture

```text
Massive Internet Text
          ↓
Next Token Prediction
          ↓
Learn General Language Patterns
          ↓
Prompt Model With Task
          ↓
Perform Many NLP Tasks
```

---

# Most Important Contribution

The paper helped establish the modern LLM idea:

```text
general-purpose language models
```

instead of:
- one model per task
- task-specific architectures.

This became the foundation for:
- GPT-3
- InstructGPT
- ChatGPT
- modern generative AI.

# GPT vs GPT-2

GPT-2 is the successor to the original GPT.

| Feature | GPT (2018) | GPT-2 (2019) |
|---|---|---|
| Paper | *Improving Language Understanding by Generative Pre-Training* | *Language Models are Unsupervised Multitask Learners* |
| Organization | OpenAI | OpenAI |
| Architecture | Transformer decoder | Larger Transformer decoder |
| Training Objective | next-token prediction | next-token prediction |
| Attention | causal self-attention | causal self-attention |
| Direction | left-to-right | left-to-right |
| Main Goal | pretrained + fine-tuning | zero-shot multitask learning |

---

# GPT (Original GPT)

Main idea:

```text
Pretrain language model
→ Fine-tune for downstream tasks
```

GPT introduced:
- generative pretraining
- transfer learning for NLP

---

# GPT Workflow

## Step 1 — Pretraining

Train on:
- unlabeled text

using:

```text
next-word prediction
```

Example:

```text
"The cat sat on the"
→ predict "mat"
```

---

## Step 2 — Fine-Tuning

Then fine-tune for:
- sentiment
- QA
- entailment
- classification

using labeled datasets.

---

# GPT Architecture

GPT uses:
- Transformer decoder only
- causal mask
- autoregressive generation

Meaning:

```text
future tokens hidden
```

Example:

```text
The → cat → sat → on → mat
```

---

# GPT-2

GPT-2 scaled GPT dramatically.

Main idea:

```text
Large language models can perform many tasks
WITHOUT supervised fine-tuning.
```

---

```text
task-specific fine-tuning
```

---

## GPT-2

Could often do tasks using:

```text
prompting only
```

without additional training.
