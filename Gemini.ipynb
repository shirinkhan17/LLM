
# GPT vs Gemini Architecture

| Area | GPT | Gemini |
|---|---|---|
| Main Input | text only | text + image + audio + video |
| Embeddings | text embeddings only | multimodal embeddings |
| Transformer | text Transformer | shared multimodal Transformer |
| Attention | text self-attention | multimodal self-attention |
| Generation | autoregressive decoding | autoregressive multimodal decoding |
| Core Idea | language modeling | unified multimodal reasoning |

---

# GPT Architecture

```text
Text
  ↓
Tokenizer
  ↓
Text Embeddings
  ↓
Decoder Transformer
(causal self-attention)
  ↓
Next-token prediction
  ↓
Generated Text
```

GPT processes:
- only text tokens
- only text attention

---

# Gemini Architecture

```text
Text / Image / Audio / Video
                ↓
Modality Encoders
                ↓
Multimodal Embeddings
                ↓
Shared Multimodal Transformer
(multimodal self-attention)
                ↓
Autoregressive Decoder
(next-token prediction)
                ↓
Generated Output
```

Gemini processes:
- text embeddings
- image embeddings
- audio embeddings
- video embeddings

inside ONE shared Transformer.

---

through shared multimodal attention.

# High-Level Architecture

```text
Input Modalities
(text / image / audio / video / code)
                ↓
Modality Encoders
                ↓
Embeddings
(text embeddings + visual embeddings + audio embeddings)
                ↓
Shared Multimodal Transformer
(self-attention + multimodal reasoning + fusion)
                ↓
Autoregressive Decoder
(next-token prediction)
                ↓
Generated Output
(text / code / reasoning response)


Example:
Image + Question
        ↓
Vision Encoder + Text Tokenizer
        ↓
Visual Embeddings + Text Embeddings
        ↓
Shared Multimodal Transformer
        ↓
Cross-modal Attention & Reasoning
        ↓
Autoregressive Text Generation
        ↓
Answer
```

# 1. Multimodal Inputs

Gemini accepts:
- text tokens
- image embeddings

```text
token-like representations
```

# 2. Modality Encoders

Different modalities first pass through encoders.

Examples:

| Modality | Encoder |
|---|---|
| Text | tokenizer |
| Image | vision encoder |
| Audio | audio encoder |
| Video | temporal/video encoder |

The encoders convert inputs into embeddings.

---

# 3. Shared Multimodal Transformer

The embeddings are processed together using a large Transformer.

This is the core of Gemini.

Unlike older systems:

```text
Vision model + language model
```

Gemini jointly trains modalities together.

This is called:

```text
native multimodality
```
