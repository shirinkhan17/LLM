# GPT-3 vs GPT-4

| Feature | GPT-3 | GPT-4 |
|---|---|---|
| Release | 2020 | 2023 |
| Organization | :contentReference[oaicite:2]{index=2} | :contentReference[oaicite:3]{index=3} |
| Architecture | decoder-only Transformer | improved decoder-only Transformer |
| Parameters | 175B | undisclosed |
| Modalities | text only | text + image |
| Reasoning Ability | strong | much stronger |
| Few-Shot Learning | introduced powerful few-shot learning | greatly improved |
| Instruction Following | moderate | very strong |
| RLHF | limited/early | heavily used |
| Context Length | 2K | much larger |
| Hallucinations | more frequent | reduced |
| Coding Ability | good | much stronger |
| Safety Alignment | weaker | significantly improved |
| Long Context Handling | limited | strong |
| Conversational Quality | good | much more natural |
| Inference Cost | very high | even higher |
| Main Innovation | emergent few-shot learning | reasoning + multimodal AI |

# 1. GPT-3

Main idea:

```text
Scale language models massively
```

GPT-3 introduced:
- powerful few-shot learning
- in-context learning
- emergent abilities

Example:

```text
English → French examples
```

The model infers tasks directly from prompts.

---

## GPT-3 Strengths

- strong text generation
- few-shot learning
- language understanding

---

## GPT-3 Weaknesses

- hallucinations
- weaker reasoning
- less reliable instruction following
- text-only model

---

# 2. GPT-4

Main idea:

```text
Improve reasoning, alignment, and multimodal capability
```

GPT-4 improves:
- reasoning
- coding
- safety
- conversation quality
- instruction following

# Final Summary

GPT-3 introduced powerful few-shot learning and emergent language abilities through massive scaling, while GPT-4 significantly improved reasoning, instruction following, safety, long-context understanding, and multimodal capability using stronger alignment and post-training techniques.
